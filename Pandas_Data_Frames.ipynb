```python
#Q.26:
course_df = pd.DataFrame({
    "Student": ["Amit", "Ravi", "Priya", "Neha", "Rahul"],
    "Course": ["Python", "Java", "DBMS", "Python", "Java"],
    "Score": [85, 78, 92, 88, 75]
})

# Display column names
print(course_df.columns)

# Retrieve Student column
print(course_df["Student"])
# Create DataFrame with Student and Score
student_score_df = course_df[["Student", "Score"]]
print(student_score_df)
# Add Batch column
course_df["Batch"] = "B1"

print(course_df)
```

```python
#Q.27:
travel_df = pd.DataFrame({
    "Name": ["Amit", "Ravi", "Priya", "Neha", "Rahul"],
    "Destination": ["Delhi", "Mumbai", "Pune", "Jaipur", "Agra"]
})

# Add Code using np.arange(5)
travel_df["Code"] = np.arange(5)

print("After adding Code:")
print(travel_df)

# Create Series with values only at indexes 1 and 4
travel_code = pd.Series(
    [101, 104],
    index=[1, 4]
)

# Replace Code column
travel_df["Code"] = travel_code

print("\nAfter replacing Code:")
print(travel_df)
```

```python
#Q.28:
school_df = pd.DataFrame({
    "Student": ["Amit", "Ravi", "Priya", "Neha", "Rahul"],
    "Class": [10, 10, 11, 11, 10],
    "Section": ["A", "B", "A", "B", "A"]
})

# Request House even though it does not exist
new_school_df = school_df.reindex(
    columns=["Student", "Class", "Section", "House"]
)

print("Before assigning House:")
print(new_school_df)

# Assign one common value to House
new_school_df["House"] = "Red House"

print("\nAfter assigning House:")
print(new_school_df)
```

```python
#Q.29:
inventory_df = pd.DataFrame({
    "Item": ["Laptop", "Mouse", "Keyboard", "Monitor", "Printer"],
    "Quantity": [10, 25, 15, 8, 5],
    "Price": [50000, 800, 1500, 12000, 7000]
})

# Create partially indexed Series
stock_series = pd.Series(
    [20, 30, 15],
    index=[0, 2, 4]
)

# Add Series as new column
inventory_df["Stock"] = stock_series

print("After adding Stock:")
print(inventory_df)

# Delete Stock column
inventory_df.drop("Stock", axis=1, inplace=True)

print("\nAfter deleting Stock:")
print(inventory_df)
```

```python
#Q.30:
# Read copied table
lab_df = pd.read_clipboard()

# Display complete DataFrame
print("DataFrame:")
print(lab_df)

# Display column names
print("\nColumn names:")
print(lab_df.columns)

# Retrieve a single-word column
print("\nFirst column:")
print(lab_df[lab_df.columns[0]])

# Retrieve a column whose name contains a space
print("\nColumn containing space:")
print(lab_df[lab_df.columns[1]])

# Create smaller DataFrame using selected columns
small_df = lab_df[[lab_df.columns[0], lab_df.columns[1], lab_df.columns[2]]]

print("\nSmaller DataFrame:")
print(small_df)

# Add a new whole column
lab_df["NewColumn"] = "Test"

print("\nAfter adding new column:")
print(lab_df)

# Delete the new column
lab_df.drop("NewColumn", axis=1, inplace=True)

print("\nAfter deleting new column:")
print(lab_df)
```
