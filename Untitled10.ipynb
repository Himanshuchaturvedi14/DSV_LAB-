```python
import pandas as pd
from pandas import Series

sales = Series(
    [450, 750, 620, 300],
    index=['Pen', 'Notebook', 'Bag', 'Pencil']
)

# Display values
print(sales.values)

# Display index
print(sales.index)

# Select one label
print(sales['Notebook'])

# Filter sales above 500
print(sales[sales > 500])

# Check whether a label exists
print('Bag' in sales.index)

# Convert Series to dictionary
sales_dict = sales.to_dict()
print(sales_dict)
```

```python
data = {'A': 15, 'B': 35, 'C': 25}

s = Series(data, index=['C', 'A', 'D', 'B'])

s.name = 'Observations'
s.index.name = 'Category'
print("Complete Series:")
print(s)

# (a)
print("\n(a) Index Order:")
print(s.index)

# (b)
print("\n(b) Label containing NaN:")
print(s[s.isnull()].index)

# (c)
print("\n(c) Values greater than 20:")
print(s[s > 20])

# (d)
print("\n(d) notnull():")
print(pd.notnull(s))


print("\n(e) Series Name:")
print(s.name)
print("Index Name:")
print(s.index.name)
```
