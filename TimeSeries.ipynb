```python
import pandas as pd
from datetime import datetime
```

```python
dt = datetime(2026, 2, 15, 10, 30)
print("Year:", dt.year)
print("Month:", dt.month)
print("Day:", dt.day)
print("Hour:", dt.hour)
print("Minute:", dt.minute)
```

```python
dt1 = datetime(2026, 2, 15, 10, 0)
dt2 = datetime(2026, 2, 15, 12, 30)
diff = dt2 - dt1
print("Hours:", diff.seconds // 3600)
print("Minutes:", (diff.seconds % 3600) // 60)
```

```python
dates = ["2026-02-01", "2026-02-05", "2026-02-10"]
date_values = pd.to_datetime(dates)
print(date_values)
```

```python
dates = ["2026-02-01", "invalid", "2026-02-05", None]
date_values = pd.to_datetime(dates, errors="coerce")
print(date_values)
print(date_values.isna())
```

```python
dates = pd.date_range("2026-02-01", periods=6, freq="D")
daily_sales = pd.Series([100, 120, 150, 130, 180, 200],
 index=dates)
```

```python
df = pd.DataFrame({
    "Current Sales": daily_sales,
    "Previous Day Sales": daily_sales.shift(1)
})
print(df)
```

```python
df = pd.DataFrame({
    "Current Sales": daily_sales,
    "Next Day Sales": daily_sales.shift(-1)
})
print(df)
```

```python
change = daily_sales - daily_sales.shift(1)
print(change)
```

```python
shifted_sales = daily_sales.shift(2, freq="D")
print(daily_sales)
print(shifted_sales)
```
